# **Práctica 9. Mejorar calidad de imagen a color**

**Importación de librerías**

```python
import cv2
import matplotlib.pyplot as plt
```

Se realiza la importación de librerías necesarias para llevar a cabo la práctica. OpenCV (cv2) se encarga del procesamiento y las operaciones realizadas en las imágenes, mientras que matplotlib se utiliza para lograr visualizar los resultados.

# **Parte I.- Iluminación de Imagen**

```python
# Cargar las imágenes seleccionadas
imagen_bgr = cv2.imread("/content/Im004_ACRIMA.jpg")
imagen_bgr_2 = cv2.imread("/content/mcar-1.jpg")
imagen_bgr_3 = cv2.imread("/content/1-001.jpg")

imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)
imagen_rgb_2 = cv2.cvtColor(imagen_bgr_2, cv2.COLOR_BGR2RGB)
imagen_rgb_3 = cv2.cvtColor(imagen_bgr_3, cv2.COLOR_BGR2RGB)

# Convertir las imágenes a HSV
imagen_hsv = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2HSV)
imagen_hsv_2 = cv2.cvtColor(imagen_bgr_2, cv2.COLOR_BGR2HSV)
imagen_hsv_3 = cv2.cvtColor(imagen_bgr_3, cv2.COLOR_BGR2HSV)

# Separar canales H, S y V de cada imagen
h, s, v = cv2.split(imagen_hsv)
h_2, s_2, v_2 = cv2.split(imagen_hsv_2)
h_3, s_3, v_3 = cv2.split(imagen_hsv_3)

# Ecualizar solamente el canal de valor (brillo) de cada imagen
v_eq = cv2.equalizeHist(v)
v_eq_2 = cv2.equalizeHist(v_2)
v_eq_3 = cv2.equalizeHist(v_3)

# Combinar canales H y S con el nuevo canal V ecualizado
imagen_hsv_eq = cv2.merge([h, s, v_eq])
imagen_hsv_eq_2 = cv2.merge([h_2, s_2, v_eq_2])
imagen_hsv_eq_3 = cv2.merge([h_3, s_3, v_eq_3])

# Convertir de nuevo a BGR y luego a RGB para visualización
imagen_eq_bgr = cv2.cvtColor(imagen_hsv_eq, cv2.COLOR_HSV2BGR)
imagen_eq_bgr_2 = cv2.cvtColor(imagen_hsv_eq_2, cv2.COLOR_HSV2BGR)
imagen_eq_bgr_3 = cv2.cvtColor(imagen_hsv_eq_3, cv2.COLOR_HSV2BGR)

imagen_eq_rgb = cv2.cvtColor(imagen_eq_bgr, cv2.COLOR_BGR2RGB)
imagen_eq_rgb_2 = cv2.cvtColor(imagen_eq_bgr_2, cv2.COLOR_BGR2RGB)
imagen_eq_rgb_3 = cv2.cvtColor(imagen_eq_bgr_3, cv2.COLOR_BGR2RGB)

# Mostrar todas las imágenes originales y mejoradas
plt.figure(figsize=(12, 12))

plt.subplot(3, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen 1 original")
plt.axis("off")

plt.subplot(3, 2, 2)
plt.imshow(imagen_eq_rgb)
plt.title("Imagen 1 mejorada (Ecualización V)")
plt.axis("off")

plt.subplot(3, 2, 3)
plt.imshow(imagen_rgb_2)
plt.title("Imagen 2 original")
plt.axis("off")

plt.subplot(3, 2, 4)
plt.imshow(imagen_eq_rgb_2)
plt.title("Imagen 2 mejorada (Ecualización V)")
plt.axis("off")

plt.subplot(3, 2, 5)
plt.imshow(imagen_rgb_3)
plt.title("Imagen 3 original")
plt.axis("off")

plt.subplot(3, 2, 6)
plt.imshow(imagen_eq_rgb_3)
plt.title("Imagen 3 mejorada (Ecualización V)")
plt.axis("off")

plt.show()
```

Para el desarrollo de esta práctica se cargaron 3 imágenes distintas por medio de la función "cv2.imread()" y se pasaron a RGB con la función "cv2.cvtColor()" como en prácticas pasadas. Después, cada imagen se convirtió al espacio de color HSV para luego separar cada canal con "cv2.split()", en donde los canales corresponden al tono, saturación y valor (brillo).

Posteriormente, se aplicó una ecualización al canal V con "cv2.equalizeHist()". Lo que hace esa función es básicamente redistribuir los niveles de intensidad en el histograma correspondiente a lo largo de todo el rango de 0 a 255, mejorando el contraste y aclarando zonas oscuras.

Con el canal V ecualizado, se volvieron a unir los canales H y S originales con el nuevo canal V con "cv2.merge()". Por último, las imágenes se pasaron de HSV a BGR y luego a RGB para así visualizarlas correctamente, mostrando cada una en forma de comparación con ayuda de matplotlib.
